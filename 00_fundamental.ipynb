* Reshaping, it is redining shape
* view: return certain shape but keep memory as orginal memory
* stacking:combine multibple tensors on top'(concat)
* squeezing': remove one dimention from a tensor
'*tensors': add one dim to a tensor

* permute': return **view** swapped in certain way


```python
import torch

t = torch.arange(12)
print(t)
t_view = t.view(3,4 )
print(t,t_view)
t_view[0,0] = 999
print(t,t_view)
#updating t or t_view is same as one is copy of the other

a = torch.tensor([1, 2, 3])
b = torch.tensor([4, 5, 6])
c = torch.tensor([4,5,6])
stacked_tensor = torch.stack([a, b,c], dim=0)
print(stacked_tensor)

t_extra = torch.tensor([[[5, 10, 15],[4,5,6],[9,9,8]]])

t_squeezed = t_extra.squeeze()
print(t_extra.shape,t_squeezed.shape)

t_1d = torch.tensor([10, 20, 30])

t_unsqueezed = t_1d.unsqueeze(0)
t_un = t_unsqueezed.unsqueeze(1)
t_unn = t_unsqueezed.unsqueeze(2)
print(t_unsqueezed,t_unsqueezed.shape)
print(t_un,t_un.shape)
print(t_unn,t_unn.shape)
img_tensor = torch.randn(3, 28, 28)
img_permuted = img_tensor.permute(1, 0, 2)
print(img_tensor.shape)
print(img_permuted.shape)
```

TO reduce **randomness** we use **random seed**

```python
## ge(nerated randomness
import torch

a = torch.rand(3,4)
b = torch.rand(3,4)

print(a)
print(b)
print(a==b)
```

```python
import torch

rand_seed = 42.9
torch.manual_seed(rand_seed)
c = torch.rand(3,4)
rand_seed = 42.8
torch.manual_seed(rand_seed)
d = torch.rand(3,4)
print(c)
print(d)
print(c==d)
```
